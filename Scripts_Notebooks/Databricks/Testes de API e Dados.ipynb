```python
# Teste de Conxao com o olinda PTAX

import requests

url = "https://olinda.bcb.gov.br/olinda/servico/PTAX/versao/v1/odata/CotacaoDolarDia(dataCotacao=@dataCotacao)?@dataCotacao='09-18-2026'&$format=json"

response = requests.get(url, timeout=10)
print(response.status_code)
print(response.text)
```

```python
# Texte de verificacao de conteudo do dado

import json

dados = response.json()
print(json.dumps(dados, indent=2, ensure_ascii=False))
```

```python
# Teste de dados com Multiplos periodos

url_periodo = (
    "https://olinda.bcb.gov.br/olinda/servico/PTAX/versao/v1/odata/"
    "CotacaoDolarPeriodo(dataInicial=@dataInicial,dataFinalCotacao=@dataFinalCotacao)"
    "?@dataInicial='09-15-2026'&@dataFinalCotacao='09-18-2026'&$format=json"
)

response_periodo = requests.get(url_periodo, timeout=10)
print(response_periodo.status_code)

dados_periodo = response_periodo.json()
print(json.dumps(dados_periodo, indent=2, ensure_ascii=False))
```

```python
# Teste de dados com inclusão do boletim diário para análise do padrão da API

import requests
import json

url_boletim = (
    "https://olinda.bcb.gov.br/olinda/servico/PTAX/versao/v1/odata/"
    "CotacaoMoedaPeriodo(moeda=@moeda,dataInicial=@dataInicial,dataFinalCotacao=@dataFinalCotacao)"
    "?@moeda='USD'&@dataInicial='09-17-2026'&@dataFinalCotacao='09-21-2026'"
    "&$format=json"
    "&$select=paridadeCompra,paridadeVenda,cotacaoCompra,cotacaoVenda,dataHoraCotacao,tipoBoletim"
)

response_boletim = requests.get(url_boletim, timeout=10)
print(response_boletim.status_code)

dados_boletim = response_boletim.json()
print(json.dumps(dados_boletim, indent=2, ensure_ascii=False))
```
